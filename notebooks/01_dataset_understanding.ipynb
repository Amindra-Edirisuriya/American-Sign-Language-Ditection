```python
total_files = 0
all_zero_files = 0
nonzero_files = 0
missing_column_files = 0
missing_values = 0
    
```

```python
for file in dataset_path.rglob("*.csv"):
    total_files += 1

    data = pd.read_csv(file)

    if "Visibility" not in data.columns:
        missing_column_files += 1
        continue

    visibility = data["Visibility"]

    missing_values += visibility.isna().sum()

    if visibility.dropna().eq(0).all():
        all_zero_files += 1
    else:
        nonzero_files += 1
```

```python
print("Total files:", total_files)
print("All-zero files:", all_zero_files)
print("Files with non-zero values:", nonzero_files)
print("Files missing Visibility:", missing_column_files)
print("Total missing Visibility values:", missing_values)
```

```python
files_with_outside_coordinates = 0
```

```python
for file in dataset_path.rglob("*.csv"):

    data = pd.read_csv(file)

    outside = (
        (data["X"] < 0) |
        (data["X"] > 1) |
        (data["Y"] < 0) |
        (data["Y"] > 1)
    )

    if outside.any():
        files_with_outside_coordinates += 1
```

```python
print("Files with coordinates outside 0-1:",
      files_with_outside_coordinates)
```

```python
"""
#### Normalized Coordinate Range Analysis
/*
**Objective:** Investigate X and Y coordinate values outside the expected normalized image range of 0–1.

**Results:**

| Condition | Count |
|---|---:|
| X below 0 | 2 |
| X above 1 | 23 |
| Y below 0 | 34 |
| Y above 1 | 67 |
| Total outside-range coordinate values | 126 |
| Files containing outside-range coordinates | 50 |

**Conclusion:**

A total of 50 samples out of 64,376 contain at least one X or Y coordinate outside the 0–1 range. These values may represent hand landmarks outside the image boundaries. Further investigation is required before deciding whether any correction or removal is necessary.
"""
```

```python
outside_by_class = {}
```

```python
for class_folder in dataset_path.iterdir():

    if class_folder.is_dir():

        class_name = class_folder.name
        affected_files = 0

        for file in class_folder.glob("*.csv"):

            data = pd.read_csv(file)

            outside = (
                (data["X"] < 0) |
                (data["X"] > 1) |
                (data["Y"] < 0) |
                (data["Y"] > 1)
            )

            if outside.any():
                affected_files += 1

        outside_by_class[class_name] = affected_files
```
